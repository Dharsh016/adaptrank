```python
# notebooks/umap_profiles.ipynb  (run as notebook, not script)

import os
import pickle
import numpy as np
import pandas as pd
import umap
import plotly.express as px
from pathlib import Path

# Set working directory to project root (parent of notebooks/)
project_root = Path.cwd()
if project_root.name == 'notebooks':
    project_root = project_root.parent
os.chdir(project_root)

with open("data/processed/learner_profiles.pkl","rb") as f:
    profiles = pickle.load(f)

df_info = pd.read_csv("data/raw/studentInfo.csv")

sids = list(profiles.keys())
vecs = np.array([profiles[s] for s in sids])

reducer = umap.UMAP(n_components=2, random_state=42, n_neighbors=15)
coords = reducer.fit_transform(vecs)

plot_df = pd.DataFrame({'x':coords[:,0], 'y':coords[:,1], 'id_student':sids})
plot_df = plot_df.merge(
    df_info[['id_student','final_result']].drop_duplicates('id_student'),
    on='id_student', how='left')

fig = px.scatter(plot_df, x='x', y='y', color='final_result',
                 title='Learner profile space (UMAP)',
                 opacity=0.5, width=800, height=600)
fig
```
